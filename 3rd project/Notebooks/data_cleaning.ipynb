# Data Cleaning 


```python
# Create a copy of original dataset


df_clean = df.copy()

print("Backup created successfully")
print("Original Records : " , len(df))
print("Working Records : " , len(df_clean))

```

Remove duplicate records

```python
# Check duplicates

print("Before Cleaning : " , len(df_clean))

# Remove duplicate rows


df_clean = df_clean.drop_duplicates()

print("After cleaning:", len(df_clean))
```

```python
# Duplicate IDs

print("Duplicates Ids : " , df_clean["ID"].duplicated().sum() )

# We should not automatically remove duplicate IDs without inspecting them because two records might have different property details.
```

Handle missing values

```python
# Check missing values
print(df_clean.isnull().sum())
```

```python
# Numerical columns
num_cols = df_clean.select_dtypes(include=np.number).columns

for col in num_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())


# Categorical columns
cat_cols = df_clean.select_dtypes(include=["object","str"]).columns

for col in cat_cols:
    if df_clean[col].isnull().any():
        df_clean[col] = df_clean[col].fillna(
            df_clean[col].mode()[0]
        )

        
```

```python

print("Total missing values:", df_clean.isnull().sum().sum())
```

Clean categorical values
Sometimes the same category is written differently, for example:

```python
# Clean text columns
cat_cols = df_clean.select_dtypes(include=["object","str"]).columns

for col in cat_cols:
    df_clean[col] = df_clean[col].str.strip()
    df_clean[col] = df_clean[col].str.title()

print("Categorical values cleaned!")
```

Detect outliers using IQR
Outliers are unusually high or low values.

For example, most properties might cost between ₹20 lakh and ₹100 lakh, while a few might cost ₹500 lakh.

We will use the IQR method to identify them.

```python
# Function to detect outliers

def detect_outliers(column):

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)
    ]

    print("Column:", column)
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print("Outliers:", len(outliers))

    return outliers
```

```python
# Now apply it:

price_outliers = detect_outliers("Price_in_Lakhs")

size_outliers = detect_outliers("Size_in_SqFt")

sqft_outliers = detect_outliers("Price_per_SqFt")
```
