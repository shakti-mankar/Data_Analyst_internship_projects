## 5. Data Quality Check

### 5.1 Missing Value Analysis

```python
# step 1 missing values

df.isnull().sum()

```

```python
# step 2 missing values analysis for return_3yr

df[df['returns_3yr'].isnull()][
    ['scheme_name','fund_age_yr','category','sub_category','returns_1yr','returns_3yr','returns_5yr']
]
```

```python
# step 3 missing values analysis for return_5yr

df[df['returns_5yr'].isnull()][
    ['scheme_name','fund_age_yr','category','sub_category','returns_1yr','returns_3yr','returns_5yr']
]
```

```python
# duplicates

df.duplicated().sum()

```

```python
# Calculate missing percentages

missing_count = df.isnull().sum()

missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Values' : missing_count,
    'Missing percentage' : missing_percentage
})

missing_summary[missing_summary['Missing Values'] > 0 ]
```

```python
#the age of funds with missing 5-year returns

df[df['returns_5yr'].isnull()]['fund_age_yr'].value_counts().sort_index()
```

```python
df[df['returns_5yr'].isnull()][
    ['fund_age_yr','returns_1yr','returns_3yr']
].describe()
```

### Missing Value Treatment

```python
# The dataset contains 21 missing values in returns_3yr (2.58%) and
# 167 missing values in returns_5yr (20.52%).

# The missing values are concentrated among relatively young funds.
# Since these funds may not have sufficient historical data to calculate
# 3-year or 5-year returns, the missing values are retained as NaN rather
# than being replaced with zero or an estimated value.

# During analysis, funds will only be included in calculations involving
# a specific return period when the corresponding return data is available.
```
