### Missing Value Treatment

```python
# The dataset contains 21 missing values in returns_3yr (2.58%) and
# 167 missing values in returns_5yr (20.52%).

# The missing values are concentrated among relatively young funds.
# Since these funds may not have sufficient historical data to calculate
# 3-year or 5-year returns, the missing values are retained as NaN rather
# than being replaced with zero or an estimated value.

# During analysis, funds will only be included in calculations involving
# a specific return period when the corresponding return data is available.
```

```python
# # Checking the unique values in performance metric columns

for col in ['sortino','alpha','sd','beta','sharpe']:
    print("\n", col)
    print(df[col].unique()[:30])
```

```python
for col in ['sortino','alpha','sd','beta','sharpe']:
    print(col , ":" , df[col].map(type).value_counts())
```

```python
# Checking for non-numeric values

performance_cols = ['sortino', 'alpha', 'sd', 'beta', 'sharpe']

for col in performance_cols:
    numeric_values =  pd.to_numeric(df[col], errors='coerce')

    print(col , "-> Non_numerice values:" , numeric_values.isna().sum())
```

```python
# Identify the actual non-numeric values

for col in performance_cols:
    numeric_values = pd.to_numeric(df[col], errors='coerce')
    
    non_numeric = df.loc[numeric_values.isna(), col].unique()
    
    print(f"\n{col}")
    print(non_numeric)
```

```python
# Replace '-' with NaN and convert performance metrics to numeric

for col in performance_cols:
    df[col] = df[col].replace('-' , np.nan)

    df[col] = pd.to_numeric(df[col], errors='coerce')
```

### Missing Value Treatment

The dataset contains missing values in historical return and performance
metric fields. Missing values in `returns_3yr` and `returns_5yr` are retained
because the corresponding historical periods may not be available for all
funds.

The `-` values found in the performance metric columns were treated as
unavailable values and converted to NaN. These missing values were not
replaced with zero because an unavailable metric does not imply a value of
zero.

The missing performance metrics affect only a small portion of the dataset,
so the records are retained and missing values will be handled appropriately
during the ranking and analysis stages.
