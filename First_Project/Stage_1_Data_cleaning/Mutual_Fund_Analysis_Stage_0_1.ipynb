## 5. Data Cleaning

### 5.1 Missing Value Analysis

```python
# Count missing values
missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Values': missing_count,
    'Missing Percentage': missing_percentage
})

missing_summary[missing_summary['Missing Values'] > 0].sort_values(
    'Missing Values', ascending=False
)
```

```python
# Investigate missing 3-year and 5-year returns
df[df['returns_3yr'].isnull()][
    ['scheme_name', 'fund_age_yr', 'category', 'sub_category',
     'returns_1yr', 'returns_3yr', 'returns_5yr']
].head(20)
```

```python
df[df['returns_5yr'].isnull()][
    ['scheme_name', 'fund_age_yr', 'category', 'sub_category',
     'returns_1yr', 'returns_3yr', 'returns_5yr']
].head(20)
```

```python
# Missing 5-year returns by fund age
df[df['returns_5yr'].isnull()]['fund_age_yr'].value_counts().sort_index()
```

**Missing-value treatment decision:** Missing historical return values are retained as `NaN` rather than being replaced with zero, because a missing return does not mean that the fund earned 0%. Similarly, unavailable risk/performance metrics are retained as missing values and will be handled explicitly during later ranking and analysis.

- **Performance Metrices :** `sortino`, `alpha`, `sd`, `beta`, `sharpe`

```python
# Performance metrics expected to be numeric
performance_cols = ['sortino', 'alpha', 'sd', 'beta', 'sharpe']

for col in performance_cols:
    print(f"\n{col}")
    print("Sample values:", df[col].unique()[:20])
```

```python
# Count values that cannot be converted to numeric
for col in performance_cols:
    numeric_values = pd.to_numeric(df[col], errors='coerce')
    print(col, "-> Non-numeric values:", numeric_values.isna().sum())
```

### 5.4 Numeric Standardization

The performance metric columns were stored as text even though most values were numeric. The source data also contains `'-'` in these fields. The `'-'` values are treated as unavailable (`NaN`) rather than as zero, and the columns are then converted to numeric types.

```python
# Replace unavailable '-' values with NaN and convert performance metrics to numeric
for col in performance_cols:
    df[col] = df[col].replace('-', np.nan)
    df[col] = pd.to_numeric(df[col], errors='coerce')

df[performance_cols].info()
```

```python
# Verify missing values created by the unavailable '-' entries
df[performance_cols].isnull().sum()
```

The categorical checks are used to identify unexpected spelling/case variations or invalid ranges. The observed values are retained because they are consistent with the supplied dataset.

### 5.6 Outlier Investigation

Outliers are investigated rather than automatically deleted. An unusually high or low financial metric is not necessarily a data error. Values are retained unless there is evidence that they are incorrect.

```python
numeric_cols = [
    'min_sip', 'min_lumpsum', 'expense_ratio', 'fund_size_cr', 'fund_age_yr',
    'sortino', 'alpha', 'sd', 'beta', 'sharpe', 'risk_level', 'rating',
    'returns_1yr', 'returns_3yr', 'returns_5yr'
]

# Minimum and maximum values
df[numeric_cols].agg(['min', 'max']).T
```

```python
# Count negative values
df[numeric_cols].lt(0).sum()
```

```python
# Investigate unusually large absolute beta values
df.loc[
    df['beta'].abs() > 5,
    ['scheme_name', 'category', 'sub_category', 'beta', 'sd', 'sharpe',
     'sortino', 'alpha', 'returns_1yr', 'returns_3yr', 'returns_5yr']
]
```

```python
# Investigate unusually large absolute 1-year returns
df.loc[
    df['returns_1yr'].abs() > 50,
    ['scheme_name', 'category', 'sub_category', 'returns_1yr',
     'returns_3yr', 'returns_5yr', 'risk_level', 'sd', 'beta', 'sharpe']
]
```

```python
# Investigate unusually high standard deviation
df.loc[
    df['sd'] > 30,
    ['scheme_name', 'category', 'sub_category', 'sd',
     'returns_1yr', 'returns_3yr', 'returns_5yr', 'beta', 'sharpe']
]
```

**Outlier decision:** The extreme observations identified above are retained. The analysis found no evidence from the supplied data alone that these observations are data-entry errors. They will be considered during interpretation and later scoring rather than being arbitrarily removed.

### 5.7 Missing Value Treatment

#### Treatment decisions

- `returns_3yr` and `returns_5yr`: retain missing values as `NaN` because unavailable historical periods should not be treated as 0% return.
- `sortino`, `alpha`, `sd`, `beta`, and `sharpe`: retain `NaN` where the original dataset contained `'-'`, because unavailable metrics should not be converted to zero.
- No blanket `fillna(0)` operation is used.
- Records are retained so useful information in other columns is not unnecessarily discarded.

# Summary 

```python
summary = pd.DataFrame({
    'Mean': df[main_numeric_cols].mean(),
    'Median': df[main_numeric_cols].median(),
    'Mode': df[main_numeric_cols].mode().iloc[0],
    'Minimum': df[main_numeric_cols].min(),
    'Maximum': df[main_numeric_cols].max(),
    'Std Dev': df[main_numeric_cols].std()
})

summary
```

```python
summary_rounded = summary.round(2)

summary_rounded
```

What do these statistics tell you about the typical fund?

```python
# comparison between mean and median

comparison = pd.DataFrame({
    'Mean': df[main_numeric_cols].mean(),
    'Median': df[main_numeric_cols].median()
})

comparison['Mean - Median'] = (
    comparison['Mean'] - comparison['Median']
)

comparison.round(2)
```

### Mean vs Median Insights

The mean represents the average value, while the median represents the middle value. When they are close, the data is generally balanced. A large difference between them indicates that some unusually high or low values are affecting the average.

In this dataset, fund size has a much higher mean than median, showing that a few very large funds increase the average. For most return measures, the mean and median are relatively close, suggesting that the typical return is not heavily affected by extreme values.
